```python
#first we try with the brute force obe by one jayenge
#because we tracing each and every element so time and space complexity is more
#time complexity For n elements, the number of comparisons is roughly:o(O(n²)) as we have loop in loop nxn
nums=[1,2,3,7,8,1]
is_duplicate=False

for i in range(len(nums)):
  for j in range(i+1,len(nums)):
    if nums[i]==nums[j]:
      is_duplicate=True
      break

  if is_duplicate:
    break

print(is_duplicate)

#now we will try optimising it-- we have sets for that
nums=[1,2,3,7,8,1]
seen=set()

for num in nums:
  if num in seen:
    print(True)
    break

  seen.add(num)
else:
    print(False)
```

```python
#best time to seel and buy stockes
prices=[7,2,5,3,4,6,8]
max_profit=0

for i in range(len(prices)):
  for j in range(1+1, len(prices)):
    profit=prices[j]-prices[i]
  if profit>max_profit:
    max_profit=profit

print(max_profit)
```
